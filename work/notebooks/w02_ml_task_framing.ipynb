## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task type: Scoring / Ranking.**

I want to give each content item an opportunity score and rank pages by how strongly they may deserve CTR or engagement review. This is a scoring/ranking problem because the main decision is which pages a content or SEO reviewer should investigate first, rather than assigning each page to a simple yes/no class.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target / proxy: CTR and engagement opportunity.**

I would score how much a content item may deserve CTR or engagement review using observed performance signals such as impressions, clicks, CTR, average position, sessions, and engagement-related measures. The opportunity itself is not a directly observed label, so this is a decision-support proxy rather than a ground-truth outcome. I will keep the proxy separate from the observed metrics used to construct it and avoid claiming that the score proves a page will improve after a change.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success metric: Precision@K.**

I would use precision@K to measure how many of the top K pages identified by the score are judged to be genuine review opportunities. This matches the decision because a reviewer has limited time and will usually inspect only the highest-ranked pages. A higher precision@K would mean that more of the pages at the top of the queue are useful for review. This metric requires an observed or manually reviewed opportunity label before it can be measured honestly.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule such as “review every page with CTR below X” would be simple, but it would ignore other important signals such as search position, impressions, sessions, and engagement. A page with low CTR may be in a difficult search position, while another page with a similar CTR may have stronger visibility and represent a different opportunity. Because several signals need to be considered together, a scoring approach can provide more useful decision-support than one fixed threshold. The result should be treated as directional and used to prioritize review, not as proof that a page will improve after a change.
