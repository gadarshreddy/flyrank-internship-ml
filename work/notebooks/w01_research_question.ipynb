## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
print("Decision-support framing completed.")
```

**Decision:** Decide which visible pages should be reviewed first for possible CTR or engagement improvements.

**Who acts on it:** A content or SEO reviewer would use the ranked results to decide which pages to investigate first.

**Action:** The reviewer could inspect a selected page and consider actions such as reviewing its metadata, intent match, snippet structure, on-page engagement, or simply monitoring it.

**Cost of a wrong recommendation:** A wrong recommendation could cause the reviewer to spend limited time on a page that is not a meaningful opportunity, while a more useful page is delayed. Low-volume pages could also produce noisy signals and lead to unnecessary work.
